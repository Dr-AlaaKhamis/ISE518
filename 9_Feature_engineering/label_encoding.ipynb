# Label Encoding for Predictive Maintenance

This notebook demonstrates how categorical maintenance data can be converted into numerical labels using Python and `scikit-learn`.

**Scenario:** Historical maintenance records contain a categorical variable called `Failure_Mode`. We will encode these text labels into numbers that can be used in an analytics workflow.

> **Important:** The encoded numbers are labels, not quantities. For example, a code of `3` does not mean a failure mode is three times more severe than a code of `1`.

## 1. Import the required libraries

```python
# pandas is used to create and manipulate tabular data
import pandas as pd

# LabelEncoder converts categorical labels into integer codes
# !pip install -U scikit-learn
from sklearn.preprocessing import LabelEncoder
```

## 2. Create a simple maintenance dataset

Each row represents a historical maintenance observation and records the observed failure mode.

```python
# Define historical failure-mode observations
data = {
    "Failure_Mode": [
        "Bearing Wear",
        "Electrical Fault",
        "Lubrication Issue",
        "Bearing Wear",
        "Overheating",
        "Bearing Wear",
        "Overheating"
    ]
}

# Convert the dictionary into a pandas DataFrame
df = pd.DataFrame(data)

# Display the original data
df
```

## 3. Apply label encoding

`fit_transform()` performs two operations:
1. **Fit:** learns the unique categories in the data.
2. **Transform:** replaces each category with its corresponding integer code.

```python
# Create a LabelEncoder object
encoder = LabelEncoder()

# Learn the failure-mode categories and encode them
df["Failure_Mode_Encoded"] = encoder.fit_transform(df["Failure_Mode"])

# Compare the original categories with their encoded values
df
```

## 4. Examine the learned mapping

The encoder stores the categories it learned. The following code displays the category-to-number mapping.

```python
# Create a readable mapping from category names to encoded values
mapping = pd.DataFrame({
    "Failure_Mode": encoder.classes_,
    "Encoded_Value": encoder.transform(encoder.classes_)
})

mapping
```

## 5. Decode the numerical labels

`inverse_transform()` converts the integer codes back to their original categorical labels. This reinforces that the integers are simply representations of categories.

```python
# Convert the encoded values back to the original failure-mode labels
df["Decoded_Failure_Mode"] = encoder.inverse_transform(
    df["Failure_Mode_Encoded"]
)

df
```

## 6. Key takeaway for predictive maintenance

- Maintenance datasets often contain categorical information such as **failure mode, asset type, manufacturer, or maintenance action**.
- Machine-learning workflows often require categorical information to be represented numerically.
- Label encoding assigns an integer to each category.
- The assigned integers **do not represent magnitude, distance, severity, or rank** unless the categories genuinely have an order.

### Modeling note
`LabelEncoder` in scikit-learn is primarily intended for encoding a **target variable (`y`)**. For nominal categorical **input features (`X`)**, techniques such as `OneHotEncoder` are often more appropriate because integer labels may accidentally suggest an ordering to some models.
