# The 7-Step Procedure

Companion code for the *Predictive Analytics Lab* (section 2). Run the cell below.

```python
# The 7-step procedure in miniature
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

# 1. Characterize the problem: binary classification (malignant vs benign), target recall > 0.95
# 2. Collect data
data = load_breast_cancer()
X, y = data.data, data.target
print("Samples, features:", X.shape)
# 3. Massage the data: stratified train/test split (scaling happens inside the pipeline)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
# 4. Know your enemy: class balance and feature ranges
print("Class counts:", np.bincount(y), " feature range:", X.min().round(2), "to", X.max().round(1))
# 5. Torture the data until it confesses: train a model
model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000)).fit(Xtr, ytr)
# 6. Evaluate the confession
print(classification_report(yte, model.predict(Xte), target_names=data.target_names))
# 7. Deploy: the trained pipeline scores new samples
print("Prediction for a new sample:", data.target_names[model.predict(Xte[:1])[0]])
```
