```python
# Back-of-the-envelope Transformer arithmetic
import numpy as np

for name, L, D, V in [("Original", 12, 1024, 32000), ("GPT-2", 48, 1600, 50257), ("GPT-3", 96, 12288, 50257)]:
    print(f"{name:8s} ~{(12 * L * D * D + V * D) / 1e9:7.2f} B parameters  (12 L D^2 + V D)")

img, patch = 224, 16
print("ViT tokens:", (img // patch) ** 2, " values per patch:", patch * patch * 3)

x = np.array([2.0, -1.0, 0.5, 3.0])
print("LayerNorm:", ((x - x.mean()) / x.std()).round(3), "  RMSNorm:", (x / np.sqrt((x ** 2).mean())).round(3))

E, A, shared = 128, 8, 1
print(f"MoE: parameters x{E + shared}, compute x{A + shared}, active fraction {(A + shared) / (E + shared):.1%}")

vocab = ["need", "want", "cat", "the"]; logits = np.array([2.0, 0.5, -1.0, 1.2])
p = np.exp(logits) / np.exp(logits).sum()
print("P(next token | 'Attention is all you'):", dict(zip(vocab, p.round(2))))
```
