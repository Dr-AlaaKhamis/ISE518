```python
# Trend, seasonality, and autocorrelation of the airline passengers series
import numpy as np
import matplotlib.pyplot as plt
AIR = [112,118,132,129,121,135,148,148,136,119,104,118,115,126,141,135,125,149,170,170,158,133,114,140,145,150,178,163,172,178,199,199,184,162,146,166,171,180,193,181,183,218,230,242,209,191,172,194,196,196,236,235,229,243,264,272,237,211,180,201,204,188,235,227,234,264,302,293,259,229,203,229,242,233,267,269,270,315,364,347,312,274,237,278,284,277,317,313,318,374,413,405,355,306,271,306,315,301,356,348,355,422,465,467,404,347,305,336,340,318,362,348,363,435,491,505,404,359,310,337,360,342,406,396,420,472,548,559,463,407,362,405,417,391,419,461,472,535,622,606,508,461,390,432]
y = np.array(AIR, dtype=float); t = np.arange(len(y))

trend = np.convolve(y, np.ones(12) / 12, mode="same")          # 12-month moving average
season = np.array([np.mean((y - trend)[6:-6][(np.arange(len(y) - 12) + 6) % 12 == m]) for m in range(12)])
def acf(x, k): x = x - x.mean(); return np.sum(x[k:] * x[:-k]) / np.sum(x * x)

fig, ax = plt.subplots(1, 3, figsize=(13, 3.2))
ax[0].plot(t, y, label="data"); ax[0].plot(t[6:-6], trend[6:-6], label="trend"); ax[0].legend(); ax[0].set_title("Trend")
ax[1].bar(range(12), season); ax[1].set_title("Average seasonal effect by month")
ax[2].bar(range(1, 31), [acf(y, k) for k in range(1, 31)]); ax[2].set_title("Autocorrelation")
plt.show()
print("Lag-1 autocorrelation:", round(acf(y, 1), 3), "  lag-12:", round(acf(y, 12), 3))
```
