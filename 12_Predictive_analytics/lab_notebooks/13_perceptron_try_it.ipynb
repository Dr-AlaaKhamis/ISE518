```python
# Perceptron learning rule on OR (converges) and XOR (never converges)
import numpy as np

def train_perceptron(X, d, w, theta, eta=0.1, epochs=20):
    for epoch in range(1, epochs + 1):
        errors = 0
        for x, target in zip(X, d):
            y = int(x @ w >= theta)          # fire if the weighted sum reaches the threshold
            e = target - y
            w = w + eta * e * x              # w_i <- w_i + eta * e * x_i
            theta = theta - eta * e          # theta <- theta - eta * e
            errors += abs(e)
        print(f"epoch {epoch}: errors={errors}, w={np.round(w, 2)}, theta={theta:.2f}")
        if errors == 0:
            print("Converged!"); return w, theta
    print("Did not converge: the classes are not linearly separable.")
    return w, theta

X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
print("OR gate"); train_perceptron(X, np.array([0, 1, 1, 1]), np.array([0.3, -0.1]), 0.2)
print("\nXOR gate"); train_perceptron(X, np.array([0, 1, 1, 0]), np.array([0.3, -0.1]), 0.2, epochs=8)
```
