# Neural Network Training and Backpropagation

Companion code for the *Predictive Analytics Lab* (section 15). Run the cell below.

```python
# Backpropagation on the slides' 2-2-1 network (sigmoid, target 0.5, eta 0.5)
import numpy as np

sig = lambda z: 1 / (1 + np.exp(-z))
x = np.array([0.35, 0.9]); t = 0.5; eta = 0.5
W1 = np.array([[0.1, 0.4],     # column 0 feeds Z1 (w1, w2), column 1 feeds Z2 (w3, w4)
               [0.8, 0.6]])
W2 = np.array([0.3, 0.9])      # w5, w6

for it in range(1, 4):
    z = sig(x @ W1); y = sig(z @ W2)          # forward pass
    dk = (y - t) * y * (1 - y)                # output error term
    dz = dk * W2 * z * (1 - z)                # hidden error terms, using the OLD output weights
    W2 = W2 - eta * dk * z                    # update output weights
    W1 = W1 - eta * np.outer(x, dz)           # update hidden weights
    print(f"iteration {it}: y={y:.4f}  delta_k={dk:.4f}  w5,w6={np.round(W2, 5)}")
print("Output after 3 updates:", round(float(sig(sig(x @ W1) @ W2)), 4))

logits = np.array([1.2, 0.3, -0.8])           # Adelie, Chinstrap, Gentoo
print("Softmax:", (np.exp(logits) / np.exp(logits).sum()).round(3))
```
