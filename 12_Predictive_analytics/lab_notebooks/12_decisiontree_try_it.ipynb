# Decision Trees

Companion code for the *Predictive Analytics Lab* (section 12). Run the cell below.

```python
# ID3-style decision tree for PlayTennis
import numpy as np
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.preprocessing import OneHotEncoder

rows = ("Sunny High Weak No|Sunny High Strong No|Overcast High Weak Yes|Rain High Weak Yes|"
        "Rain Normal Weak Yes|Rain Normal Strong No|Overcast Normal Strong Yes|Sunny High Weak No|"
        "Sunny Normal Weak Yes|Rain Normal Weak Yes|Sunny Normal Strong Yes|Overcast High Strong Yes|"
        "Overcast Normal Weak Yes|Rain High Strong No")
data = [r.split() for r in rows.split("|")]
X = [r[:3] for r in data]; y = [r[3] for r in data]

def H(labels):
    _, c = np.unique(labels, return_counts=True); p = c / c.sum()
    return -(p * np.log2(p)).sum()

print(f"Entropy of the full set: {H(y):.3f}")
for j, name in enumerate(["Outlook", "Humidity", "Wind"]):
    vals = [r[j] for r in X]
    rem = sum(vals.count(v) / len(vals) * H([yy for xx, yy in zip(vals, y) if xx == v]) for v in set(vals))
    print(f"Information gain({name}) = {H(y) - rem:.3f}")

enc = OneHotEncoder(); Xe = enc.fit_transform(X).toarray()
tree = DecisionTreeClassifier(criterion="entropy", random_state=0).fit(Xe, y)
print(export_text(tree, feature_names=list(enc.get_feature_names_out(["Outlook", "Humidity", "Wind"]))))
print("D15 (Rain, High, Weak):", tree.predict(enc.transform([["Rain", "High", "Weak"]]).toarray())[0])
```
