```python
# Convolution, ReLU, and max pooling from scratch; VGG16 parameter count
import numpy as np

img = np.zeros((8, 8)); img[:, 3:5] = 1                 # a vertical bar
k = np.array([[-1, 0, 1], [-1, 0, 1], [-1, 0, 1]])       # vertical-edge kernel

def conv2d(I, K, stride=1, pad=0):
    I = np.pad(I, pad); kh, kw = K.shape
    out = (I.shape[0] - kh) // stride + 1                # O = (W - K + 2P) / S + 1
    return np.array([[np.sum(I[i*stride:i*stride+kh, j*stride:j*stride+kw] * K)
                      for j in range(out)] for i in range(out)])

fm = conv2d(img, k); print("Feature map", fm.shape, "\n", fm)
relu = np.maximum(0, fm); print("After ReLU\n", relu)
pool = relu.reshape(3, 2, 3, 2).max(axis=(1, 3)); print("2x2 max pooling\n", pool)
print("Shapes: padding 1 ->", conv2d(img, k, pad=1).shape, " stride 2 ->", conv2d(img, k, stride=2).shape)

cfg = [64, 64, "M", 128, 128, "M", 256, 256, 256, "M", 512, 512, 512, "M", 512, 512, 512, "M"]
c, total = 3, 0
for v in cfg:
    if v != "M":
        total += 3 * 3 * c * v + v; c = v
total += 7 * 7 * 512 * 4096 + 4096 + 4096 * 4096 + 4096 + 4096 * 1000 + 1000
print(f"VGG16 parameters: {total:,}")
```
