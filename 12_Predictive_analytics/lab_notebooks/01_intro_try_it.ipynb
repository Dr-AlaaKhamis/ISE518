```python
# Classification and regression on synthetic machine data
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.model_selection import train_test_split

rng = np.random.default_rng(0)
n = 200
vib = np.r_[rng.normal(2.3, .6, n // 2), rng.normal(8.8, .9, n // 2)]
temp = np.r_[rng.normal(70, 5, n // 2), rng.normal(103, 5, n // 2)]
X = np.c_[vib, temp]
y = np.r_[np.zeros(n // 2), np.ones(n // 2)]          # 0 = healthy, 1 = faulty

# Classification: healthy or faulty?
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.3, random_state=0)
clf = LogisticRegression().fit(Xtr, ytr)
print("Classification accuracy:", clf.score(Xte, yte))

# Regression: remaining useful life (RUL) from a health index
cycles = np.arange(200)
health = 1 - (cycles / 200) ** 2 + rng.normal(0, .03, 200)
rul = 200 - cycles
reg = LinearRegression().fit(health.reshape(-1, 1), rul)
print("Predicted RUL when health = 0.5:", reg.predict([[0.5]])[0].round(1))

fig, ax = plt.subplots(1, 2, figsize=(10, 3.5))
ax[0].scatter(vib, temp, c=y, cmap="coolwarm", s=12); ax[0].set(xlabel="Vibration", ylabel="Temp", title="Classification")
ax[1].scatter(health, rul, s=8); ax[1].set(xlabel="Health index", ylabel="RUL", title="Regression")
plt.show()
```
