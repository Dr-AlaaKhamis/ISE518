Companion code for the *Predictive Analytics Lab* (section 3). Run the cell below.

```python
# Requirement 1: a pattern must exist. Requirement 3: data must be available.
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score

rng = np.random.default_rng(1)
X = rng.normal(size=(600, 5))
y_pattern = (X[:, 0] + X[:, 1] ** 2 > 1).astype(int)   # a real pattern exists
y_random = rng.integers(0, 2, 600)                     # no pattern at all
for name, y in [("pattern", y_pattern), ("random labels", y_random)]:
    acc = cross_val_score(RandomForestClassifier(n_estimators=50, random_state=0), X, y, cv=5).mean()
    print(f"{name:14s} accuracy = {acc:.2f}")

sizes = [40, 80, 160, 320, 600]
accs = [cross_val_score(RandomForestClassifier(n_estimators=50, random_state=0), X[:n], y_pattern[:n], cv=5).mean() for n in sizes]
plt.plot(sizes, accs, "o-"); plt.xlabel("Training samples"); plt.ylabel("CV accuracy"); plt.title("More data, better model")
plt.show()
```
