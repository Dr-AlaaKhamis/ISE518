```python
# Cleaning a raw sensor log, then a 60/20/20 split
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split

nan = np.nan
raw = np.array([[2.1, 68, nan], [2.4, 70, 45], [2.2, nan, 46], [25.0, 71, 47],
                [2.6, 72, 49], [8.9, 103, 76], [9.4, 106, 78], [9.1, nan, 77]])
status = ["OK", "OK", "ok", "OK", "OK", "FAIL", "Fail", "FAIL"]

X = SimpleImputer(strategy="median").fit_transform(raw)            # 1. missing values
q1, q3 = np.percentile(X[:, 0], [25, 75]); iqr = q3 - q1
keep = (X[:, 0] >= q1 - 1.5 * iqr) & (X[:, 0] <= q3 + 1.5 * iqr)   # 2. outliers
X, status = X[keep], [s for s, k in zip(status, keep) if k]
Xn = MinMaxScaler().fit_transform(X)                               # 3. normalization
y = np.array([1 if s.lower() == "fail" else 0 for s in status])    # 4. labels
print("Removed rows:", int((~keep).sum()))
print(np.round(Xn, 2)); print("labels:", y)

data = np.arange(1000)
train, rest = train_test_split(data, test_size=0.4, random_state=0)
dev, test = train_test_split(rest, test_size=0.5, random_state=0)
print("train / dev / test sizes:", len(train), len(dev), len(test))
```
