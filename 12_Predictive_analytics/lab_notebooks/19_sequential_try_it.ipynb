# Sequential Learning

Companion code for the *Predictive Analytics Lab* (section 19). Run the cell below.

```python
# Turning one sensor series into sequence-learning samples
import numpy as np
from sklearn.linear_model import Ridge

rng = np.random.default_rng(0)
series = np.sin(np.linspace(0, 20, 200)) + 0.1 * rng.normal(size=200)

def windows(x, n_in, n_out):
    X, Y = [], []
    for t in range(len(x) - n_in - n_out + 1):
        X.append(x[t:t + n_in]); Y.append(x[t + n_in:t + n_in + n_out])
    return np.array(X), np.array(Y)

for name, (n_in, n_out) in {"one-to-one": (1, 1), "many-to-one": (30, 1), "many-to-many": (30, 10)}.items():
    X, Y = windows(series, n_in, n_out)
    print(f"{name:13s} inputs {X.shape} -> targets {Y.shape}")

X, Y = windows(series, 30, 1)
model = Ridge().fit(X[:120], Y[:120].ravel())
print("many-to-one model, R^2 on later windows:", round(model.score(X[120:], Y[120:].ravel()), 3))
```
