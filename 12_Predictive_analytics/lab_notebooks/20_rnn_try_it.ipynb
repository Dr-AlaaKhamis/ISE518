# Recurrent Neural Networks (RNN)

Companion code for the *Predictive Analytics Lab* (section 20). Run the cell below.

```python
# A character-level RNN forward pass on "hello", and vanishing gradients
import numpy as np

vocab = ["h", "e", "l", "o"]; onehot = np.eye(4)
rng = np.random.default_rng(0)
Wxh, Whh, Why = rng.normal(0, .5, (3, 4)), rng.normal(0, .5, (3, 3)), rng.normal(0, .5, (4, 3))
h = np.zeros(3)
for ch, target in zip("hell", "ello"):
    h = np.tanh(Wxh @ onehot[vocab.index(ch)] + Whh @ h)     # h_t = tanh(Wxh x_t + Whh h_(t-1))
    scores = Why @ h; p = np.exp(scores) / np.exp(scores).sum()
    print(f"input '{ch}' -> hidden {h.round(2)} -> P(next) {dict(zip(vocab, p.round(2)))}  target '{target}'")

print()
for w in [0.5, 0.9, 1.0, 1.1]:
    print(f"w_hh = {w}: gradient factor after 30 steps = {w ** 30:.2e}")
```
