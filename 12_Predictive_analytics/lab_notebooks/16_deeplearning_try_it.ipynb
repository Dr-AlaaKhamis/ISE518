```python
# Rules of thumb for hidden neurons, and shallow+fat versus deep+thin networks
import numpy as np
from sklearn.datasets import load_digits
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split

ni, no, m, alpha = 10, 1, 1000, 5
print("Rule 1: n_h = m / (alpha (ni + no)) =", round(m / (alpha * (ni + no))))
print("Rule 2: n_h = 2 sqrt((no + 2) m)   =", round(2 * np.sqrt((no + 2) * m)))

X, y = load_digits(return_X_y=True); X = X / 16
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.3, random_state=0)
for name, layers in {"shallow + fat (256)": (256,), "deep + thin (32, 32, 32)": (32, 32, 32)}.items():
    net = MLPClassifier(layers, max_iter=300, random_state=0).fit(Xtr, ytr)
    params = sum(w.size for w in net.coefs_) + sum(b.size for b in net.intercepts_)
    print(f"{name:26s} params={params:6d}  test accuracy={net.score(Xte, yte):.3f}")
```
