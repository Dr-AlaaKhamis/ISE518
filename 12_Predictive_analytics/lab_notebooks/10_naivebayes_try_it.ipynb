# Naive Bayes Classifier

Companion code for the *Predictive Analytics Lab* (section 10). Run the cell below.

```python
# Gaussian Naive Bayes on the slides' machine data, plus the car-theft exercise
import numpy as np
from sklearn.naive_bayes import GaussianNB, CategoricalNB
from sklearn.preprocessing import OrdinalEncoder

X = np.array([[9.8, 105, 78], [9.2, 110, 74], [7.5, 98, 80], [8.9, 102, 72],
              [1.8, 65, 42], [2.5, 75, 48], [2.2, 70, 45], [2.8, 72, 50]])
y = np.array(["Faulty"] * 4 + ["Healthy"] * 4)
nb = GaussianNB(var_smoothing=1e-12).fit(X, y)
x = np.array([[6.0, 80, 58]])
print("Means:\n", nb.theta_.round(3)); print("Variances:\n", nb.var_.round(4))
print("Prediction:", nb.predict(x)[0], " P(class | x):", nb.predict_proba(x).round(6))

g = lambda v, m, s2: np.exp(-(v - m) ** 2 / (2 * s2)) / np.sqrt(2 * np.pi * s2)
for k, c in enumerate(nb.classes_):
    print(f"{c}: likelihood product = {np.prod(g(x[0], nb.theta_[k], nb.var_[k])):.4e}")

cars = [["Red", "Domestic", "Sports"], ["Red", "Domestic", "Sports"], ["Red", "Domestic", "Sports"],
        ["Yellow", "Domestic", "Sports"], ["Yellow", "Imported", "Sports"], ["Yellow", "Imported", "SUV"],
        ["Yellow", "Imported", "SUV"], ["Yellow", "Domestic", "SUV"], ["Red", "Imported", "SUV"], ["Red", "Imported", "Sports"]]
stolen = ["Yes", "No", "Yes", "No", "Yes", "No", "Yes", "No", "No", "Yes"]
enc = OrdinalEncoder(); Xc = enc.fit_transform(cars)
cnb = CategoricalNB(alpha=1e-10).fit(Xc, stolen)        # (almost) no smoothing, as in the slides
q = enc.transform([["Red", "Domestic", "SUV"]])
print("Red domestic SUV stolen?", cnb.predict(q)[0], dict(zip(cnb.classes_, cnb.predict_proba(q)[0].round(3))))
```
