Companion code for the *Predictive Analytics Lab* (section 4). Run the cell below.

```python
# Full versus fractional factorial designs
import itertools
import numpy as np

factors = ["A", "B", "C", "D"]
full = np.array(list(itertools.product([-1, 1], repeat=4)))
base = np.array(list(itertools.product([-1, 1], repeat=3)))
half = np.c_[base, base.prod(axis=1)]                    # 2^(4-1) design with generator D = ABC
print("Full factorial runs:", len(full), "  Fractional runs:", len(half))
print(half)

rng = np.random.default_rng(0)
def response(d):   # true model: y = 10 + 3A - 2B + 1.5AB + noise
    return 10 + 3 * d[:, 0] - 2 * d[:, 1] + 1.5 * d[:, 0] * d[:, 1] + rng.normal(0, .3, len(d))

for name, d in [("full", full), ("half", half)]:
    y = response(d)
    effects = {f: round(2 * np.mean(y * d[:, i]), 2) for i, f in enumerate(factors)}
    print(name, "main effects:", effects)

# Aliasing: in the half fraction the AB interaction column equals the CD column
print("AB aliased with CD:", np.array_equal(half[:, 0] * half[:, 1], half[:, 2] * half[:, 3]))
```
