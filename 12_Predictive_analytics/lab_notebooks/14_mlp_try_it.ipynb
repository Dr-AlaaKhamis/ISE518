# Multi-layer Perceptron (MLP)

Companion code for the *Predictive Analytics Lab* (section 14). Run the cell below.

```python
# An MLP solves XOR; compare activation functions
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neural_network import MLPClassifier

X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]]); y = np.array([0, 1, 1, 0])
mlp = MLPClassifier(hidden_layer_sizes=(4,), activation="tanh", solver="lbfgs", random_state=1, max_iter=2000).fit(X, y)
print("XOR predictions:", mlp.predict(X), " (target [0 1 1 0])")

x = np.linspace(-5, 5, 400)
acts = {"step": (x >= 0).astype(float), "sigmoid": 1 / (1 + np.exp(-x)), "tanh": np.tanh(x), "ReLU": np.maximum(0, x)}
for name, f in acts.items():
    plt.plot(x, f, label=name)
plt.ylim(-1.5, 2); plt.legend(); plt.grid(alpha=.3); plt.title("Activation functions")
plt.show()
```
