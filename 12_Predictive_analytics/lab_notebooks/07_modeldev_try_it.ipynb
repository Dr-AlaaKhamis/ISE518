```python
# No Free Lunch: compare four algorithms on three different problems
from sklearn.datasets import make_moons, make_classification, load_digits
from sklearn.model_selection import cross_val_score
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression

datasets = {"moons": make_moons(400, noise=.25, random_state=0),
            "linear": make_classification(400, n_features=20, n_informative=5, random_state=0),
            "digits": load_digits(return_X_y=True)}
models = {"Naive Bayes": GaussianNB(), "k-NN": KNeighborsClassifier(),
          "Tree": DecisionTreeClassifier(random_state=0), "Logistic": LogisticRegression(max_iter=2000)}
print(f"{'model':12s}" + "".join(f"{d:>10s}" for d in datasets))
for m, model in models.items():
    print(f"{m:12s}" + "".join(f"{cross_val_score(model, X, y, cv=5).mean():10.3f}" for X, y in datasets.values()))
print("\nNo single model wins everywhere: the No Free Lunch theorem in practice.")
```
