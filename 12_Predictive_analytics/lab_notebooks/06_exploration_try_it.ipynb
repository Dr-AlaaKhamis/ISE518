```python
# Univariate versus multivariate views, and linear separability
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles, make_blobs
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC

sets = {"blobs": make_blobs(300, centers=2, random_state=3), "rings": make_circles(300, factor=.4, noise=.08, random_state=0)}
for name, (X, y) in sets.items():
    lin = LogisticRegression().fit(X, y).score(X, y)
    rbf = SVC().fit(X, y).score(X, y)
    print(f"{name}: linear accuracy {lin:.2f}, nonlinear (RBF SVM) {rbf:.2f}")

X, y = sets["rings"]
r = np.hypot(X[:, 0], X[:, 1])
fig, ax = plt.subplots(1, 3, figsize=(12, 3.2))
ax[0].scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", s=10); ax[0].set_title("Multivariate view")
ax[1].hist([X[y == 0, 0], X[y == 1, 0]], bins=20, label=["class 0", "class 1"]); ax[1].set_title("Univariate view of x1"); ax[1].legend()
ax[2].hist([r[y == 0], r[y == 1]], bins=20); ax[2].set_title("Engineered feature: radius")
plt.show()
```
